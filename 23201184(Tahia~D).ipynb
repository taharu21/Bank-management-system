```python
import random

# Minimum 2 classes
#class 1
class Player:
    def __init__(self, name):
        self.name = name
        self.role = "Villager"
        self.alive = True
#class 2
class Game:
    def __init__(self, players):
        self.players = players

    # Wolf assign using Random
    def assign_wolf(self):
        wolves = random.sample(self.players, 2)

        for player in wolves:
            player.role = "Werewolf"

    # Night Phase
    def night(self):

        alive_players = []

        for player in self.players:
            if player.alive:
                alive_players.append(player)

        # Custom Events
        if random.randint(1, 10) == 1:
            print("Very hot weather! No attack tonight.")
            return

        victim = random.choice(alive_players)

        while victim.role == "Werewolf":
            victim = random.choice(alive_players)

        victim.alive = False

        print(victim.name, "was killed by the Werewolves.")

    # Day Phase
    def day(self):

        alive_players = []

        for player in self.players:
            if player.alive:
                alive_players.append(player)

        # Custom Events
        if random.randint(1, 10) == 2:
            print("Clear moonlight!.It's risky to attack villagers tonight.so,No attack tonight.")
            return

        suspect = random.choice(alive_players)

        suspect.alive = False

        print(suspect.name, "was voted out.")

    # Check Winner
    def check_winner(self):

        wolves = 0
        villagers = 0

        for player in self.players:

            if player.alive:

                if player.role == "Werewolf":
                    wolves += 1
                else:
                    villagers += 1

        if wolves == 0:
            return "Villagers Win"

        if wolves >= villagers:
            return "Werewolves Win"

        return None

    # Play Game
    def play(self):

        self.assign_wolf()

        while True:

            self.night()

            result = self.check_winner()

            if result:
                break

            self.day()

            result = self.check_winner()

            if result:
                break

        print("Game Over")
        print(result)

        # Save result
        file = open("result.txt", "w")

        file.write("Game Result")
        file.write(result + "  ")

        for player in self.players:

            status = "Alive"

            if not player.alive:
                status = "Dead"

            file.write(player.name + " - " +
                       player.role + " - " +
                       status + "\n")

        file.close()


# Read players from file
players = []

file = open("players.txt", "r")

for line in file:
    name = line.strip()

    if name != "":
        players.append(Player(name))

file.close()

# Start Game
game = Game(players)
game.play()
```
